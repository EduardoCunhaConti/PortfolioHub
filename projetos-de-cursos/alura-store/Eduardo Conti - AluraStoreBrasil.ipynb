### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

# 4. Produtos Mais e Menos Vendidos

```python
contagem = loja['Produto'].value_counts()

max_vendas = contagem.max()
min_vendas = contagem.min()

produtos_mais_vendidos = contagem[contagem == max_vendas]
produtos_menos_vendidos = contagem[contagem == min_vendas]

print("Produtos mais vendidos na loja 1:\n", produtos_mais_vendidos.reset_index().rename(columns={'index': 'Produto', 'count': 'Quantidade'}))
print("\nProdutos menos vendidos na loja 1:\n",produtos_menos_vendidos.reset_index().rename(columns={'index': 'Produto', 'count': 'Quantidade'}))
```

```python
contagem2 = loja2['Produto'].value_counts()

max_vendas2 = contagem2.max()
min_vendas2 = contagem2.min()

produtos_mais_vendidos2 = contagem2[contagem2 == max_vendas2]
produtos_menos_vendidos2 = contagem2[contagem2 == min_vendas2]

print("Produtos mais vendidos na loja 2:\n", produtos_mais_vendidos2.reset_index().rename(columns={'index': 'Produto', 'count': 'Quantidade'}))
print("\nProdutos menos vendidos na loja 2:\n",produtos_menos_vendidos2.reset_index().rename(columns={'index': 'Produto', 'count': 'Quantidade'}))
```

# 5. Frete Médio por Loja

```python
frete_total1 = 0
for frete in loja['Frete']:
  frete_total1 += frete
media1 = round(frete_total1 / len(loja), 2)
print("Frete médio da loja 1: ", media1)

frete_total2 = 0
for frete in loja2['Frete']:
  frete_total2 += frete
media2 = round(frete_total2 / len(loja2), 2)
print("Frete médio da loja 2: ", media2)

frete_total3 = 0
for frete in loja3['Frete']:
  frete_total3 += frete
media3 = round(frete_total3 / len(loja3), 2)
print("Frete médio da loja 3: ", media3)

frete_total4 = 0
for frete in loja4['Frete']:
  frete_total4 += frete
media4 = round(frete_total4 / len(loja4), 2)
print("Frete médio da loja 4: ", media4)
```

```python
medias = [media1, media2, media3, media4]
lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

pd.Series(medias, index=lojas).plot(kind='line', marker='o', title='Frete médio por loja')
```

## **6. Relatório de venda de unidade**

  Essa análise tem como objetivo comparar o desempenho das quatro lojas do Senhor João para identificar qual unidade apresenta menor potencial de retorno e, portanto, deve ser vendida. Para isso, foram considerados faturamento, vendas por categoria, avaliações dos clientes, produtos mais e menos vendidos e frete médio.

  Em relação ao faturamento, a Loja 1 apresenta o melhor desempenho (1.534.509,12 reais), seguida pelas Lojas 2 e 3, enquanto a Loja 4 possui o menor faturamento (1.384.497,58 reais), indicando menor geração de receita. Quanto às categorias de produtos, todas as lojas apresentam um padrão semelhante, com destaque para móveis e eletrônicos como as categorias mais vendidas, sem que a Loja 4 se sobressaia positivamente. No quesito avaliação dos clientes, a Loja 3 apresenta a melhor média (4,05), enquanto a Loja 1 possui a menor (3,98), mas as diferenças são pequenas entre as lojas. Os produtos mais e menos vendidos apresentam volumes próximos entre si, não indicando vantagem competitiva clara de nenhuma unidade. Em relação ao frete médio, a Loja 4 possui o menor custo (31,28 reais), porém esse fator isolado não é suficiente para compensar seu menor faturamento.

  Diante desse conjunto de informações, conclui-se que a Loja 4 é a melhor candidata para venda, pois apresenta o pior desempenho financeiro geral e não se destaca positivamente em nenhuma das métricas analisadas. Dessa forma, a recomendação é que o Senhor João mantenha as lojas com maior faturamento e melhor desempenho global, preservando aquelas com maior potencial de crescimento e rentabilidade.
